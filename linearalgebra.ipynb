```python
import numpy as np
import pandas as pd
```

1. Load the dataset and handle any missing data by replacing null values with the mean
value of the respective column. (Score 2)

```python
df = pd.read_csv('wineq.csv',sep=';')
df.head()
```

```python
print(df.isnull().sum())
print("  ")
df= df.fillna(df.mean(numeric_only=True))
print(df.isnull().sum())
```

There are no missing values in this dataset, so no values actually needed to be replaced.

2. Extract the following columns as vectors: alcohol,citric acid. (Score 2)

```python
alcohol = df["alcohol"].to_numpy()
citric = df["citric acid"].to_numpy()
```

```python
print("Alcohol vector : ")
print(alcohol)

print("\nCitricACid Vector: ")
print(citric)
```

3. Select two features (e.g., alcohol and density) from the dataset and calculate the
covariance matrix using np.cov(X.T), where X is the feature matrix consisting of the
selected columns.(Score 2)

```python
X = df[["alcohol","density"]]
covmatrix = np.cov(X.T)

print("Covariance Matrix: ")
print(covmatrix)
```

The diagonal values represent the variance of alcohol and density.
The off-diagonal values represent the covariance between alcohol and density.
The negative covariance indicates a slight negative relationship between alcohol and density.


4. Perform eigen decomposition on the covariance matrix you computed in question 3.
Identify and interpret the results:Identify the top 2 eigenvalues of the covariance
matrix,Identify the corresponding eigenvectors.(Score 2)

```python
eigenvalues, eigenvectors = np.linalg.eig(covmatrix)
print("Eigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```

```python
index = np.argsort(eigenvalues)[::-1]

eigenvalues = eigenvalues[index]
eigenvectors = eigenvectors[:, index]

print("Top 2 Eigenvalues:")
print(eigenvalues)

print("\nCorresponding Eigenvectors:")
print(eigenvectors)
```

The first eigenvalue is much larger than the second one. This means that most of the variation in the selected features (alcohol and density) is along the direction represented by the first eigenvector.

The first eigenvector is mainly associated with alcohol, because its value for alcohol is almost 1.

5. Which wine quality is most common in the dataset? How can you interpret the
distribution of wine quality scores? (Score 2)

```python
quality_counts = df["quality"].value_counts().sort_index()
print(quality_counts)
```

```python
print("Most common wine quality: ",quality_counts.idxmax())
print("Number of Wines: ",quality_counts.max())
```

The most common wine quality is 5, with 681 wines.



Interpretation:

The distribution is concentrated mainly around quality scores 5 and 6. Very few wines have quality scores of 3, 4, or 8. Therefore, most wines in this dataset have average or moderate quality, while very low and very high quality wines are less common.
